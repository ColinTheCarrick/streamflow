### Part 1a - Shear Diagram

You will create your shear diagram in the code block below.


```python
L = 6
Ra = 170
Rb = 410
P1 = 100
q1 = 120
q2 = 120

x = np.linspace(0, L, 500)
v = np.zeros_like(x)
for i in range(len(x)):
  if x[i] < 2:
    v[i] = Ra - (q1*x[i])
  elif x[i] < 4:
    v[i] = Ra - (q1*2) - P1
  else:
    v[i] = Ra - (q1*2) - P1 + Rb - (q2*(x[i]-4))

plt.plot(x,v)
plt.title("Shear Diagram")
plt.fill_between(x,v,color='cyan',alpha=0.2)
plt.axhline(0, color='black', linestyle='-', linewidth=1, label='y=0')

plt.xlabel("x (ft)")
plt.ylabel("V (lb)")
plt.show()
```

### Part 1b - Moment Diagram

You will create your moment diagram in the code block below.

```python
m = np.zeros_like(x)
for i in range(len(x)):
    if x[i] < 2:
        m[i] = (Ra*x[i]) - (q1*((x[i]**2)/2))
    elif x[i] < 4:
        m[i] = (Ra*x[i]) - (q1*2*(x[i] - 1)) - (P1 * (x[i] - 2))
    else:
        m[i] = (Ra*x[i]) - (q1*2*(x[i] - 1)) - (P1 * (x[i] - 2)) + (Rb*(x[i] - 4)) - (q2 * (x[i] - 4)*((x[i] - 4)/2))

plt.plot(x, m)
plt.title("Moment Diagram")
plt.xlabel("x (ft)")
plt.ylabel("M (lb-ft)")
plt.fill_between(x,m,color='cyan',alpha=0.2)
plt.axhline(0, color='black', linestyle='-', linewidth=1, label='y=0')
plt.show()
```

### Part 2 - Pasadena Precipitation Data

You will now create a graph using Pasadena Temperature Data from 2013-2014.

Create and slice the array to create the x and y-axis values. These will correspond to the following columns in the CSV linked on readthedocs:

*   x - Date
*   y_1 - High Temperature

```python
temperature_data = np.genfromtxt(
    "2013-2014-Pasadena-Temperature-Data.csv",
    delimiter=",",
    dtype=str,
    skip_header=1
)

x = temperature_data[:, 0]                  # Date
y_1 = temperature_data[:, 1].astype(float)  # High temperature
y_2 = temperature_data[:, 2].astype(float)
```

### Graphing the Data

Plot x, y_1 and y_2 with the elements shown in the instructions on the HW page.

```python
plt.figure(figsize=(10, 6))
plt.plot(x,y_1)
plt.plot(x,y_2)

plt.title("Pasadena Temperature Data 2013-2014")
plt.xlabel("Date")
plt.xticks(x[::20], rotation=45) # x[::20] shows every dates for every 20 or whatever number is used.
plt.ylabel("Degrees (Fahrenheit)")
plt.grid(True)
plt.legend(["High Temperature", "Low Temperature"])
plt.show()
```
